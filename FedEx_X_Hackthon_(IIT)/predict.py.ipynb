```python
import pandas as pd
```

```python
# Load trained model
import joblib
model = joblib.load("recovery_model.pkl")
```

```python
def assign_priority(recovery_probability):
    if recovery_probability < 0.4:
        return "HIGH"
    elif recovery_probability < 0.7:
        return "MEDIUM"
    else:
        return "LOW"
```

```python
def predict_case(case_data):
    df = pd.DataFrame([case_data])

    # Predict probability
    recovery_probability = float(model.predict_proba(df)[0][1])
    risk_score = float(1 - recovery_probability)
    priority = assign_priority(recovery_probability)

    return {
    "recovery_probability": round(recovery_probability, 3),
    "risk_score": round(risk_score, 3),
    "priority": priority
}

```

```python
# ---- Example case (simulate backend request) ----
sample_case = {
    "amount_due": 45000,
    "days_overdue": 120,
    "past_defaults": 2,
    "credit_score": 620,
    "dca_success_rate": 0.65,
    "contact_attempts": 5
}
```

```python
result = predict_case(sample_case)
```
