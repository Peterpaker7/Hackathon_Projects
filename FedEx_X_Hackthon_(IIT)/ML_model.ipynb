```python
import pandas as pd
```

```python
dataset=pd.read_csv('synthetic_cases.csv')
```

```python
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['amount_due', 'days_overdue', 'past_defaults',
       'credit_score', 'dca_success_rate', 'contact_attempts']]
```

```python
dependent=dataset[['recovered']]
```

```python
from sklearn.model_selection import train_test_split as tts
x_train,x_test,y_train,y_test=tts(independent,dependent,test_size=1/3,random_state=0)
```

```python
y_test
```

```python
from sklearn.linear_model import LogisticRegression
reg=LogisticRegression(random_state=0)
reg.fit(x_train,y_train)
```

```python
y_pred=reg.predict(x_test)
```

```python
from sklearn.metrics import roc_auc_score
auc = roc_auc_score(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report
clf_report=classification_report(y_test,y_pred)
```

```python
print(clf_report)
```

```python
# Save model
import joblib
joblib.dump(reg, "recovery_model.pkl")

print("Model trained and saved successfully!")
```
